```python
#1
# ตรวจสอบว่าคะแนนถูกต้องหรือไม่
def is_valid_score(score):
    if score >= 0 and score <= 100:
        return True
    else:
        return False

# แปลงคะแนนเป็นเกรด
def to_grade(score):
    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "F"

# แสดงผลคะแนน
def print_result(student_id, score):
    if is_valid_score(score):
        grade = to_grade(score)
        print("รหัสนักศึกษา:", student_id)
        print("คะแนน:", score)
        print("เกรด:", grade)
    else:
        print("รหัสนักศึกษา:", student_id)
        print("คะแนนไม่ถูกต้อง")

# รับข้อมูลจากผู้ใช้
student_id = input("กรอกรหัสนักศึกษา: ")
score = float(input("กรอกคะแนน: "))

# เรียกใช้ฟังก์ชันแสดงผล
print_result(student_id, score)
```

```python

# ฟังก์ชันหาค่าเฉลี่ย
def safe_average(total, count):
    if count <= 0:
        return None
    return round(total / count, 2)

total = 0
count = 0

# รับค่า PM2.5 จำนวน 5 วัน
day1 = float(input("วันที่ 1: "))
day2 = float(input("วันที่ 2: "))
day3 = float(input("วันที่ 3: "))
day4 = float(input("วันที่ 4: "))
day5 = float(input("วันที่ 5: "))

# ตรวจสอบข้อมูลวันที่ 1
if day1 >= 0 and day1 <= 500:
    total = total + day1
    count = count + 1

# ตรวจสอบข้อมูลวันที่ 2
if day2 >= 0 and day2 <= 500:
    total = total + day2
    count = count + 1

# ตรวจสอบข้อมูลวันที่ 3
if day3 >= 0 and day3 <= 500:
    total = total + day3
    count = count + 1

# ตรวจสอบข้อมูลวันที่ 4
if day4 >= 0 and day4 <= 500:
    total = total + day4
    count = count + 1

# ตรวจสอบข้อมูลวันที่ 5
if day5 >= 0 and day5 <= 500:
    total = total + day5
    count = count + 1

# หาค่าเฉลี่ย
average = safe_average(total, count)

# แสดงผลลัพธ์
if average == None:
    print("ไม่มีข้อมูลที่ถูกต้อง")
else:
    print("ค่าเฉลี่ย PM2.5 =", average)
```
