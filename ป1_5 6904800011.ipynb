```python
#1
# ใช้ list เพราะมีลำดับ และเป็นตัวเลขที่ต้องวนลูปคำนวณ
scores = [78, 92, 65, 88, 45, 95, 72]

if len(scores) > 0:  # ตรวจก่อนว่า list ไม่ว่าง จะได้ไม่หารด้วย 0
    print("จำนวนผู้เข้าสอบ :", len(scores))
    print("คะแนนสูงสุด :", max(scores))
    print("คะแนนต่ำสุด :", min(scores))
    avg = sum(scores) / len(scores)
    print("ค่าเฉลี่ย :", round(avg, 2))

    over80 = 0       # ตัวนับ ต้องสร้างไว้นอกลูป
    fail = []        # list ว่างไว้เก็บคนที่ไม่ผ่าน สร้างไว้นอกลูปเหมือนกัน
    for s in scores:
        if s >= 80:
            over80 = over80 + 1
        if s < 50:
            fail.append(s)

    print("ได้ 80 ขึ้นไป :", over80, "คน")
    print("คะแนนที่ไม่ผ่านเกณฑ์ 50 :", fail)
else:
    print("ไม่มีข้อมูล")
```

```python
#2
ids = ["C001", "C002", "C001", "C003", "C002", "C004"]

# ใช้ set นับจำนวนที่ไม่ซ้ำ เพราะ set ไม่เก็บค่าซ้ำ
unique_count = len(set(ids))
print("จำนวนรายการทั้งหมด :", len(ids))
print("จำนวนรหัสที่ไม่ซ้ำ :", unique_count)
print("จำนวนรายการที่ซ้ำ :", len(ids) - unique_count)

# ต้องเก็บลำดับที่เจอครั้งแรก เลยใช้ set() ตรง ๆ ไม่ได้
# วนลูปเอง ถ้ายังไม่มีใน list ใหม่ค่อย append
unique_ids = []
for i in ids:
    if i not in unique_ids:
        unique_ids.append(i)
print("รหัสที่ไม่ซ้ำ (เรียงตามที่เจอก่อน) :", unique_ids)
```

```python
#3
payments = ["cash", "credit", "cash", "transfer", "cash", "credit"]

# ใช้ dict เพราะต้องเก็บคู่ "วิธีชำระเงิน -> จำนวนครั้ง"
counter = {}
for p in payments:
    counter[p] = counter.get(p, 0) + 1   # ถ้ายังไม่เคยเจอให้เริ่มที่ 0 แล้วบวก 1

# แสดงผลเรียงตามชื่อวิธี
for method in sorted(counter):
    print(method, ":", counter[method], "ครั้ง")

# หาวิธีที่ใช้มากที่สุด
best_method = ""
best_count = 0
for method, n in counter.items():
    if n > best_count:
        best_count = n
        best_method = method
print("ใช้มากที่สุด :", best_method, "(", best_count, "ครั้ง )")
```

```python

#4
orders = [
    {"id": 1001, "customer": "Alice", "city": "Bangkok", "product": "Notebook", "qty": 2, "price": 120},
    {"id": 1002, "customer": "Bob", "city": "Bangkok", "product": "Pen", "qty": 5, "price": 15},
    {"id": 1003, "customer": "Charlie", "city": "Chiang Mai", "product": "Backpack", "qty": 1, "price": 850},
    {"id": 1004, "customer": "Diana", "city": "Phuket", "product": "Notebook", "qty": 3, "price": 120},
    {"id": 1005, "customer": "Eve", "city": "Bangkok", "product": "Pen", "qty": 10, "price": 15},
    {"id": 1006, "customer": "Frank", "city": "Phuket", "product": "Backpack", "qty": 1, "price": 850}
]

# 1. ยอดขายรวมทั้งหมด
total = 0

for o in orders:
    total = total + (o["qty"] * o["price"])

print("1) ยอดขายรวมทั้งหมด :", total, "บาท")


# 2. ยอดขายแยกตามสินค้า
product_sales = {}

for o in orders:
    product = o["product"]
    money = o["qty"] * o["price"]

    if product in product_sales:
        product_sales[product] = product_sales[product] + money
    else:
        product_sales[product] = money

print("2) ยอดขายแยกตามสินค้า :")

for product in product_sales:
    print(product, "-", product_sales[product], "บาท")


# 3. จำนวนลูกค้าที่ไม่ซ้ำ
customers = []

for o in orders:
    if o["customer"] not in customers:
        customers.append(o["customer"])

print("3) จำนวนลูกค้าที่ไม่ซ้ำ :", len(customers), "คน")


# 4. 3 อันดับลูกค้ายอดขายสูงสุด
customer_sales = []

for o in orders:
    money = o["qty"] * o["price"]
    name = o["customer"]

    customer_sales.append([money, name])

customer_sales.sort(reverse=True)

print("4) 3 อันดับลูกค้ายอดขายสูงสุด :")

for i in range(3):
    print(customer_sales[i][1], "-", customer_sales[i][0], "บาท")
```

```python
#5
# โครงสร้างที่เลือก : dict ที่คีย์คือสัปดาห์ ค่าคือ set ของรหัสนักศึกษาที่มาเรียน
# เหตุผล
#  - dict : แต่ละสัปดาห์มีชื่อของตัวเอง ("week1", "week2", ...)
#  - set  : ในหนึ่งสัปดาห์คนหนึ่งมาได้แค่ครั้งเดียว ไม่ต้องมีค่าซ้ำ และไม่สนลำดับ
#  - set ทำให้ตอบข้อ (2) ได้ด้วย len() และตอบข้อ (3) ได้ด้วยเครื่องหมาย - กับ |
attendance = {
    "week1": {"C001", "C002", "C003"},
    "week2": {"C002", "C003", "C005"},
    "week3": {"C001", "C003"},
}

# รายชื่อนักศึกษาทั้งหมดในห้อง
all_students = {"C001", "C002", "C003", "C004", "C005", "C006"}

# ตอบข้อ (3) ใครที่ไม่เคยเข้าเรียนเลย
# รวมคนที่เคยมาอย่างน้อยหนึ่งครั้งไว้ใน came ก่อน (ใช้ | ทีละสัปดาห์)
came = set()   # ต้องใช้ set() ห้ามเขียน {} เพราะจะได้ dict ว่าง
for week in attendance:
    came = came | attendance[week]

never_came = all_students - came   # อยู่ในห้อง แต่ไม่เคยมา
print("นักศึกษาที่เคยมาอย่างน้อย 1 ครั้ง :", sorted(came))
print("นักศึกษาที่ไม่เคยเข้าเรียนเลย :", sorted(never_came))
```
